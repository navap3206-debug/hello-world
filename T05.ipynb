## Calcula el R2 e interpreta los signos de los betas.

Escalamos TODO el data frame para este ejercisio

```python
#ESCALADA DE TODO EL DT
dt_all = dt.copy()
scaler = StandardScaler()
X_escalado = scaler.fit_transform(dt_all[['cyl','disp', 'hp', 'drat','wt', 'qsec', 'gear', 'carb']])
X_escalado_all= pd.DataFrame(X_escalado, columns=['cyl','disp', 'hp', 'drat','wt', 'qsec', 'gear', 'carb'])
X_escalado_all['vs'] = dt_all['vs'] #LOS CATEGORICOS LOS REINCORPORAMOS
X_escalado_all['am'] = dt_all['am']
vec_y_all = dt['mpg']
print(X_escalado_all.describe())
```

```python
X_escalado_all
```

Realizamos la regresion de TODO EL dataframe para conseguir R squared y coeficiented




```python
#REGRESIÓN Y R2
import statsmodels.api as SM
X2 = SM.add_constant(X_escalado_all) #CONCADENAR
model = SM.OLS(vec_y_all,X2) #MODELO MINIMOS CUADRADOS ORDINARIOS
results1 = model.fit() #OPTIMIZA
print(results1.summary())#RESUME
```

```python
# División 40% entrenamiento / 60% prueba
X_train, X_test, y_train, y_test = train_test_split(dt[['cyl','disp', 'hp', 'drat','wt', 'qsec', 'vs', 'am', 'gear', 'carb']], dt['mpg'], train_size=0.4, random_state=42)
X_train.info()
```

```python
X_escalado_train = X_escalado_train.reset_index(drop=True) #TENEMOS UN DESMADRE EN LOS INIcES ASÍ QUE RESETEAMOS LOS INDICES
X_escalado_test = X_escalado_test.reset_index(drop=True)
X_escalado_train
```

ENTRENAMOMS CON LA PARTE TRAIN DE DT


```python
X2 = SM.add_constant(X_escalado_train)
model = SM.OLS(y_train.reset_index(drop=True), X2)
results2 = model.fit()
print(results2.summary())
```

Comprobamos r-squared de test

```python
X2_test = SM.add_constant(X_escalado_test)
X_test_predicr = results2.predict(X2_test)
r2_score(y_test, X_test_predicr)
```

```python
ridge = Ridge(alpha=1)
ridge.fit(X_escalado_train, y_train)
y_train_pred = ridge.predict(X_escalado_train)
print(f'Resultados RIDGE(L2) -> intercepto: {ridge.intercept_}, coeficientes: {ridge.coef_}')
print(f' R-SQUARED3 {r2_score(y_train, y_train_pred)}')

```
