```python
import pandas as pd
from google.colab import files
```

PREPARACIÓM DE LOS DATOS


```python
dt = files.upload()
df = pd.read_csv('feo.csv', encoding='latin-1')

df.head(100)
```

TRANFORMAMOS A BINARIO

```python
df['KAHLUA'] = (df['SABOR'] == 'Kahlua').astype(int)
df['HORCHATA'] = (df['SABOR'] == 'Horchata').astype(int)
df['HIELO'] = (df['HIELO'] == 'con').astype(int)
df['TAMAÑO'] = (df['TAMAÑO'] == 'grande').astype(int)

```

```python
df.drop(columns=['SABOR'], inplace=True)
```

```python

df.head(8)
```

```python
import numpy as np
from sklearn.linear_model import LinearRegression
#TRANSFORMAMOS A UNA MATRIZ MANIPULABLE

X_matrix = df[['TAMAÑO', 'HIELO', 'KAHLUA', 'HORCHATA']].to_numpy()
Y_vec = df['SCORE'].to_numpy()

lr = LinearRegression(fit_intercept= True)
lr.fit(X_matrix, Y_vec)
lr.intercept_

```

```python
lr.intercept_
```

```python
lr.coef_
```

```python
#con summary
import statsmodels.api as SM
X2 = SM.add_constant(X_matrix)
model = SM.OLS(Y_vec,X2)
results = model.fit()
print(results.summary())
```

PARTE 2 (MODIFICAMOS LOS CATEGORICOS DE TIPO 1,-1


```python
df['HIELO'] = df['HIELO'].replace(0,-1)
df['TAMAÑO'] = df['TAMAÑO'].replace(1,-1)
df['TAMAÑO'] = df['TAMAÑO'].replace(0,1)
df
```

```python
X_matrix_2 = df[['TAMAÑO', 'HIELO', 'KAHLUA', 'HORCHATA']].to_numpy()
Y_vec = df['SCORE'].to_numpy()
```

```python
#con summary
import statsmodels.api as SM
X2 = SM.add_constant(X_matrix_2) #CONCADENAR
model = SM.OLS(Y_vec,X2) #MODELO MINIMOS CUADRADOS ORDINARIOS
results_2 = model.fit() #OPTIMIZA
print(results_2.summary())#RESUME
```

```python
print(results.params)
print(results_2.params)
```
