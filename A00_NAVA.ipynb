#REGRESIONES


```python

B0 = float(input())
B1 = float(input())
```

```python
m = 0
```

```python
pts = [[1,1.1],[2,2],[3,2.9],[4,4],[6,6.2],[8,8],[10,9.9]]
```

```python
for i in pts:
  m += ((B0 + B1*(i[0]))- i[1])**2
```

```python
m
```

#MATRICES

```python
#NEANDERTAL:
import numpy as np
```

```python
#Creo arrays de X / Y, además de [m,b] osea betas
X_matrix = np.array([[1, pt[0]] for pt in pts])
Y_matrix = np.array([[pt[1]] for pt in pts])
B_matrix = np.array([[B0], [B1]])

print("X:", X_matrix)
print("Y:", Y_matrix)
```

```python
matrix = (X_matrix @ B_matrix) - Y_matrix
matrix
```

```python
#Un error basado en las beta antes propuestas
ans = matrix.T @ matrix
ans
```

#SKLEARN

```python
import numpy as np
from sklearn.linear_model import LinearRegression

```

```python
lr = LinearRegression(fit_intercept= False)
lr.fit(X_matrix, Y_matrix)
lr.intercept_
```

```python
lr.coef_ #Solución de la regresión
```

```python
import matplotlib.pyplot as plt

x_values = [pt[0] for pt in pts]
y_values = [pt[1] for pt in pts]

b0 = lr.coef_[0, 0]
b1 = lr.coef_[0, 1]

x_line = np.linspace(min(x_values), max(x_values), 100)
y_line = b0 + b1 * x_line

plt.figure(figsize=(10, 6))
plt.scatter(x_values, y_values, label='Original Data Points', color='blue')
plt.plot(x_line, y_line, label=f'Regression Line (y = {b0:.2f} + {b1:.2f}x)', color='red')

plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Data Points and Linear Regression Line')
plt.legend()
plt.grid(True)
plt.show()
```
