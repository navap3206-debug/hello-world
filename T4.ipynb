## Modelos penalizados

Hasta ahora la función de costo que usamos para decidir qué tan bueno es nuestro modelo al momento de ajustar es:

$$ \text{RSS} = \sum_{i=1}^n e_i^2 = \sum_{i=1}^n (y_i - \hat{y_i})^2 $$

Dado que los errores obtenidos son una combinación de sesgo y varianza, puede ser que se sesgue un parámetro para minimizar el error. Esto significa que el modelo puede decidir que la salida no sea una combinación de los factores, sino una fuerte predilección sobre uno de los factores solamente.

E.g. se quiere ajustar un modelo

$$ \hat{z} = \hat{\beta_0} + \hat{\beta_1} x + \hat{\beta_2} y $$

Se ajusta el modelo y se decide que la mejor decisión es $\hat{\beta_1} = 10000$ y $\hat{\beta_2}=50$. Considera limitaciones de problemas reales:
- Quizás los parámetros son ajustes de maquinaria que se deben realizar para conseguir el mejor producto posible, y que $10000$ sea imposible de asignar.
- Quizás los datos actuales están sesgados y sólo hacen parecer que uno de los factores importa más que el otro.

Una de las formas en las que se puede mitigar este problema es penalizando a los parámetros del modelo, cambiando la función de costo:

$$ \text{RSS}_{L2} = \sum_{i=1}^n e_i^2  + \lambda \sum_{j=1}^p \hat{\beta_j}^2 $$

El *L2* significa que se está agregando una penalización de segundo orden. Lo que hace esta penalización es que los factores ahora sólo tendrán permitido crecer si hay una reducción al menos proporcional en el error (sacrificamos sesgo, pero reducimos la varianza).

Asimismo, existe la penalización *L1*

$$ \text{RSS}_{L1} = \sum_{i=1}^n e_i^2  + \lambda \sum_{j=1}^p |\hat{\beta_j}| $$

A las penalizaciones *L2* y *L1* se les conoce también como Ridge y Lasso, respectivamente.

Para realizar una regresión con penalización de Ridge o de Lasso usamos el objeto `Ridge(alpha=?)` o `Lasso(alpha=?)` en lugar de `LinearRegression()` de `sklearn`.

Utiliza el dataset de publicidad (Advertising.csv) y realiza 3 regresiones múltiples:

$$ \text{sales} = \beta_0 + \beta_1 (\text{TV}) + \beta_2 (\text{radio}) + \beta_3 (\text{newspaper}) + \epsilon $$

1. Sin penalización
2. Con penalización L2
3. Con penalización L1

¿Qué puedes observar al ajustar los valores de `alpha`?

Compara los resultados de los coeficientes utilizando valores diferentes de $\alpha$ y los $R^2$ resultantes.



```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from google.colab import files
```

```python
file = files.upload()
matrix = (pd.read_csv('Advertising.csv')).to_numpy()
```

```python
#REGRESION SIN PENALIZACIÓN
lr = LinearRegression(fit_intercept = True)
lr.fit(matrix[:,1:4], matrix[:,4])
print(lr.coef_ , lr.intercept_)
print(lr.score(matrix[:,1:4], matrix[:,4]))
```

```python
#REGRESION CON RIDGE
ridge = Ridge(alpha=10000000000000000)
ridge.fit(matrix[:,1:4], matrix[:,4])
print(ridge.coef_ , ridge.intercept_)
print(ridge.score(matrix[:,1:4], matrix[:,4]))
```

```python
#REGRESION CON LASSO
lasso = Lasso(alpha= 12)
lasso.fit(matrix[:,1:4], matrix[:,4])
print(lasso.coef_ , lasso.intercept_)
print(lasso.score(matrix[:,1:4], matrix[:,4]))
```

Después de experimentar con Ridge y Lasso, encontre que los coeficinetes de Ridge conforme aumentaba apha estos tendian a 0, y Lasso simplemente llegaban a 0. En el caso del score en ambos casos funcionan similar, los score en ridge tienden a 0 y los scores en lasso llegan a 0 pero pueden diferiri ampliamente con el de la regression tradicional.
